```python
import cv2
import time
import threading
import numpy as np
import torch
import os
import platform

from ultralytics import YOLO
import mediapipe as mp
import pyttsx3


EAR_THRESHOLD = 0.25
EAR_FRAMES = 20
MAR_THRESHOLD = 0.6
MAR_FRAMES = 15
ALERT_INTERVAL = 5

CAM_W, CAM_H = 1280, 720


device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)


MODEL_PATH = r"C:\Users\Vimal Antony\YOLO MODELS\yolo26n.pt"
assert os.path.isfile(MODEL_PATH), "yolo26n.pt not found"

model = YOLO(MODEL_PATH)
model.to(device)
model.fuse()

if device == "cuda":
    model.model.half()
    torch.backends.cudnn.benchmark = True

# =========================
# MEDIAPIPE
# =========================
mp_face_mesh = mp.solutions.face_mesh
face_mesh = mp_face_mesh.FaceMesh(
    max_num_faces=1,
    refine_landmarks=True,
    min_detection_confidence=0.6,
    min_tracking_confidence=0.6
)


engine = pyttsx3.init()
engine.setProperty("rate", 150)

def speak_async(text):
    threading.Thread(
        target=lambda: (engine.say(text), engine.runAndWait()),
        daemon=True
    ).start()

def beep():
    if platform.system() == "Windows":
        import winsound
        winsound.Beep(2200, 500)


LEFT_EYE = [33, 160, 158, 133, 153, 144]
RIGHT_EYE = [362, 385, 387, 263, 373, 380]
MOUTH = [61, 291, 81, 178, 13, 14]

def dist(a, b):
    return np.linalg.norm(np.array(a) - np.array(b))

def EAR(p, eye):
    return (dist(p[eye[1]], p[eye[5]]) +
            dist(p[eye[2]], p[eye[4]])) / (2 * dist(p[eye[0]], p[eye[3]]))

def MAR(p, mouth):
    return (dist(p[mouth[2]], p[mouth[3]]) +
            dist(p[mouth[4]], p[mouth[5]])) / (2 * dist(p[mouth[0]], p[mouth[1]]))


eye_cnt = 0
yawn_cnt = 0
last_alert = 0


cap = cv2.VideoCapture(0)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, CAM_W)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, CAM_H)
cap.set(cv2.CAP_PROP_FOURCC, cv2.VideoWriter_fourcc(*'MJPG'))

print("Press Q to quit")

while True:
    ret, frame = cap.read()
    if not ret:
        break

    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    results = model(rgb, verbose=False)[0]

    if results.boxes:
        box = results.boxes[0].xyxy[0].int().cpu().numpy()
        x1, y1, x2, y2 = box

        face = rgb[y1:y2, x1:x2]

        if face.size > 0:
            res = face_mesh.process(face)

            if res.multi_face_landmarks:
                lm = res.multi_face_landmarks[0]
                h, w, _ = face.shape
                pts = [(int(p.x*w), int(p.y*h)) for p in lm.landmark]

                ear = (EAR(pts, LEFT_EYE) + EAR(pts, RIGHT_EYE)) / 2
                mar = MAR(pts, MOUTH)

                eye_cnt = min(eye_cnt + 1, EAR_FRAMES) if ear < EAR_THRESHOLD else max(eye_cnt - 1, 0)
                yawn_cnt = min(yawn_cnt + 1, MAR_FRAMES) if mar > MAR_THRESHOLD else max(yawn_cnt - 1, 0)

                
                score = int(
                    (eye_cnt / EAR_FRAMES) * 60 +
                    (yawn_cnt / MAR_FRAMES) * 40
                )
                score = max(0, min(100, score))

                drowsy = score >= 60

                now = time.time()
                if drowsy and (now - last_alert >= ALERT_INTERVAL):
                    beep()
                    speak_async("Driver is drowsy. Please stay alert.")
                    last_alert = now

                cv2.rectangle(frame,(x1,y1),(x2,y2),(0,255,0),2)

                # Score bar
                bar_x, bar_y = 30, 120
                cv2.rectangle(frame,(bar_x,bar_y),(bar_x+300,bar_y+30),(50,50,50),2)
                cv2.rectangle(frame,(bar_x,bar_y),
                              (bar_x+int(3*score),bar_y+30),
                              (0,0,255) if score>=60 else (0,255,255),-1)

                cv2.putText(frame,f"Drowsiness Score: {score}%",
                            (bar_x,bar_y-10),
                            cv2.FONT_HERSHEY_SIMPLEX,0.8,(255,255,255),2)

                if drowsy:
                    cv2.putText(frame,"DROWSY ALERT!",
                                (450,120),
                                cv2.FONT_HERSHEY_SIMPLEX,
                                1.5,(0,0,255),4)

    cv2.imshow("YOLO26n Drowsiness Detection", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
